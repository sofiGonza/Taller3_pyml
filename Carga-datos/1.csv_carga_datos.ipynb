PANDAS
Es una herramienta gratuita para manejar información en filas y columnas, igual que una hoja de cálculo de Excel.Sirve para limpiar, filtrar y transformar datos numéricos o de texto con pocas líneas de código.Utiliza dos estructuras principales: las Series (columnas individuales) y los DataFrames (tablas completas).

```python
# ==========================================
# 1. IMPORTAR LIBRERÍAS
# ==========================================

import pandas as pd
import numpy as np


# ==========================================
# 2. CARGAR EL ARCHIVO CSV
# ==========================================

df = pd.read_csv("dataset_ventas.csv")


# ==========================================
# 3. MOSTRAR LAS PRIMERAS 5 FILAS
# ==========================================

print("Primeras 5 filas:")
display(df.head())


# ==========================================
# 4. MOSTRAR LAS ÚLTIMAS 5 FILAS
# ==========================================

print("Últimas 5 filas:")
display(df.tail())


# ==========================================
# 5. MOSTRAR 5 FILAS ALEATORIAS
# ==========================================

print("5 filas aleatorias:")
display(df.sample(n=5))


# ==========================================
# 6. INFORMACIÓN GENERAL DEL DATAFRAME
# ==========================================

print("Información general:")
df.info()


# ==========================================
# 7. ESTADÍSTICAS DESCRIPTIVAS
# ==========================================

print("Estadísticas descriptivas:")
display(df.describe())


# ==========================================
# 8. CANTIDAD DE FILAS Y COLUMNAS
# ==========================================

print("Cantidad de filas y columnas:")
print(df.shape)


# ==========================================
# 9. NOMBRES DE LAS COLUMNAS
# ==========================================

print("Nombres de las columnas:")
print(df.columns.tolist())


# ==========================================
# 10. VALORES NULOS
# ==========================================

print("Valores nulos por columna:")
print(df.isnull().sum())


# ==========================================
# 11. TIPOS DE DATOS
# 
==========================================
print("Tipos de datos:")
print(df.dtypes)
 
```
