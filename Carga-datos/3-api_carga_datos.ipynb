PANDAS
Es una herramienta gratuita para manejar información en filas y columnas, igual que una hoja de cálculo de Excel.Sirve para limpiar, filtrar y transformar datos numéricos o de texto con pocas líneas de código.Utiliza dos estructuras principales: las Series (columnas individuales) y los DataFrames (tablas completas).

```python
import pandas as pd
import numpy as np
import requests

url = "https://rickandmortyapi.com/api/character"
data = requests.get(url).json()


data_api = pd.DataFrame(data['results'])

data_api = data_api[['id', 'name', 'status', 'species', 'type', 'gender']]

print(data_api.head())
```
